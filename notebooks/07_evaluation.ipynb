# Controls, robustness and generalization

Requires real ZuCo and verified alignment. No scientific results are bundled. Run CPU tests before training. See ../README.md and ../docs/.

```python
from pathlib import Path
import os
if Path.cwd().name == 'notebooks':
    os.chdir(Path.cwd().parent)
from eeg_decoding.cli import main
MANIFEST = Path(os.environ.get('EEG_MANIFEST', 'data/manifests/combined_clean.jsonl'))
TEXT_CACHE = Path('data/processed/sentence_embeddings.npz')
CONFIG = 'configs/local.yaml'
CHECKPOINTS = Path('checkpoints')
RESULTS = Path('results')
```

```python
ARTIFACT = CHECKPOINTS/'ssl_aligned/best.pt'
main(['evaluate','--manifest',str(MANIFEST),'--artifact',str(ARTIFACT),'--text-cache',str(TEXT_CACHE),'--output',str(RESULTS/'metrics/clean.json'),'--robustness'])
for control in ['gaussian','shuffled']:
    main(['evaluate','--manifest',str(MANIFEST),'--artifact',str(ARTIFACT),'--text-cache',str(TEXT_CACHE),'--output',str(RESULTS/f'metrics/{control}.json'),'--control',control])
```

The cells above are inference controls. Also train separate Gaussian-input and shuffled-pair models (`--control` on training) under identical splits. Repeat for unseen content and both transfer directions with audited identities.
